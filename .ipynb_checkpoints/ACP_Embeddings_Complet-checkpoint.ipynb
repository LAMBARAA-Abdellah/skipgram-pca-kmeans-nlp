# Étape 2 – Visualisation des mots en 2D avec PCA (ACP)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

print('Embedding matrix shape:', np.array(embedding_matrix).shape)
print('Nombre de mots:', len(vocab))
```

## PCA from scratch (méthode classroom)

```python
def pca_from_scratch(X, k=2):
    # Centrer les données
    X_centered = X - np.mean(X, axis=0)

    # Matrice de covariance
    cov_matrix = np.cov(X_centered.T)

    # Valeurs propres / vecteurs propres
    eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

    # Trier par valeur propre décroissante
    sorted_idx = np.argsort(eigenvalues)[::-1]
    eigenvectors = eigenvectors[:, sorted_idx]

    # Garder k composantes principales
    components = eigenvectors[:, :k]

    # Projection
    X_reduced = X_centered @ components
    return X_reduced

reduced_scratch = pca_from_scratch(np.array(embedding_matrix), k=2)
print('Shape after PCA scratch:', reduced_scratch.shape)
```

## PCA sklearn

```python
pca = PCA(n_components=2)
reduced_sklearn = pca.fit_transform(embedding_matrix)
print('Shape sklearn PCA:', reduced_sklearn.shape)
```

## Visualisation 2D des mots (PCA sklearn)

```python
plt.figure(figsize=(16,12))
for i, word in enumerate(vocab):
    x, y = reduced_sklearn[i]
    plt.scatter(x, y, c='blue', s=12)
    plt.text(x+0.01, y+0.01, word, fontsize=8)

plt.title('Projection 2D des mots par PCA (ACP)')
plt.xlabel('Composante principale 1')
plt.ylabel('Composante principale 2')
plt.grid(True)
plt.show()
```
