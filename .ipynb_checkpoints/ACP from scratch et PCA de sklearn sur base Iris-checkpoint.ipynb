# Import des bibliothèques

```python
##Pandas, Numpy, Matplotlib
import pandas as ps
import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import eig
```

# Import des données

```python
#Importer les données 
#data1 = ps.read_csv('Iris.csv')
#df= data1[['SepalLengthCm','SepalWidthCm']]
#df = data1.iloc[:,1:5]
#df.head()

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

# load dataset into Pandas DataFrame
df = ps.read_csv(url, names=['sepal length','sepal width','petal length','petal width','target'])
print(df.head())
```

#  Standarisation des données

```python
from sklearn.preprocessing import StandardScaler
features = ['sepal length', 'sepal width', 'petal length', 'petal width']
# Separating out the features
x = df.loc[:, features].values
# Separating out the target
y = df.loc[:,['target']].values
# Standardizing the features
X_cr = StandardScaler().fit_transform(x)
X_cr.shape
print(X_cr)
```

# Fonction PCA

```python
#Calcul de pourcentage d'inertie
def inertie_percent(Lambda):
    L=[]
    for i in range(Lambda.shape[0]):
        L.append(Lambda[i]/sum(Lambda)*100)
    return np.array(L)

def ACP(X , num_components):
    #Centrer les données
    #X_c = X.sub(X.mean()) # Soustraire la moyenne de chaque colonne à chaque valeur : centrer les valeurs
    X_c = X-X.mean()
    #X_cr= X_c.div(X_c.std())
    #print(X_c)
    #Cov matrix
    Cov_Xc = (1/X_c.shape[0])*np.dot(X_c.T,X_c)
    #print("********** La matrice de covariance **********")
    #print (Cov_Xc)
    et= X_c.std() #ecartType
    X_cr = X_c/et
    #print("********** La matrice centré réduite **********")
    #print(X_cr)
    print("********** La matrice correlation **********")
    corr = (1/X_cr.shape[0])*np.dot(X_cr.T,X_cr)
    print(corr)
    #Les valeurs propres 
    eigen_values , eigen_vectors = np.linalg.eigh(corr)
    print("********** Les valeurs propres **********")
    print(eigen_values)
    print("********** Les vecteurs propres ********** ")
    print(eigen_vectors)
    #plt.plot(eigen_values)
    print("********** Pourcentage d'inertie **********")
    print(inertie_percent(eigen_values))
    
    sorted_index = np.argsort(eigen_values)[::-1]
    sorted_eigenvalue = eigen_values[sorted_index]
    sorted_eigenvectors = eigen_vectors[:,sorted_index]
    #plt.plot(sorted_eigenvalue)
    print("********** Pourcentage d'inertie apres trie **********")
    print(inertie_percent(sorted_eigenvalue))
    #Nombre d'axes
    eigenvector_subset = sorted_eigenvectors[:,0:num_components]
    #Projection des données sur les "num_components" premiers axes
    X_reduced = np.dot(eigenvector_subset.transpose() , X_cr.transpose() ).transpose()
    plt.plot(X_reduced[:,0],X_reduced[:,1],"*")
    return X_reduced


```

```python
 components = ACP(x,2)
```

#  PCA from sklearn

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)

principalComponents = pca.fit_transform(x)
```

#### PCA du Sklearn

```python
principalDf = ps.DataFrame(data = principalComponents
             , columns = ['principal component 1', 'principal component 2'])
principalDf.head(10)
```

## 2. Pourcentage d'inertie | Variance expliquée

#### PCA du Sklearn

```python
pca.explained_variance_ratio_
```

## 3. PLOT

#### PCA du Sklearn

```python
finalDf = ps.concat([principalDf, df[['target']]], axis = 1)
```

```python
fig = plt.figure(figsize = (8,8))
ax = fig.add_subplot(1,1,1) 
ax.set_xlabel('Principal Component 1', fontsize = 15)
ax.set_ylabel('Principal Component 2', fontsize = 15)
ax.set_title('2 component PCA', fontsize = 20)

targets = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica']
colors = ['r', 'g', 'b']
for target, color in zip(targets,colors):
    indicesToKeep = finalDf['target'] == target
    ax.scatter(finalDf.loc[indicesToKeep, 'principal component 1']
               , finalDf.loc[indicesToKeep, 'principal component 2']
               , c = color
               , s = 50)
ax.legend(targets)
ax.grid()
```
