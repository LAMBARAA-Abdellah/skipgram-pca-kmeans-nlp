```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import re
```

```python
corpus = """
Le Maroc est un pays situé à l'Afrique du nord. Sa capitale est Rabat. Cela possède une riche histoire.
Marrakech est une ville connue pour sa médina très animée. Non seulement elle est surprenante,
mais aussi le Casablanca est reconnu comme étant le centre économique du pays.
Le 'Roi' et 'l'informatique' ne se lient pas forcément, contrairement à l'information par les mots.
L'apprentissage automatique est un sous-domaine de l'intelligence artificielle qui permet
d'apprendre à partir de données.
"""

corpus = re.sub(r"[^a-zA-Zàâçéèêëîïôûùüÿñæœ’']+", " ", corpus)
words = corpus.lower().split()
print(len(words), words[:20])
```

```python
vocab = sorted(set(words))
word2idx = {w:i for i,w in enumerate(vocab)}
idx2word = {i:w for w,i in word2idx.items()}
vocab_size = len(vocab)
vocab_size
```

```python
def generate_skipgram_pairs(tokenized_text, window_size=2):
    pairs = []
    for idx, center in enumerate(tokenized_text):
        start = max(0, idx-window_size)
        end = min(len(tokenized_text), idx+window_size+1)
        for i in range(start, end):
            if i != idx:
                pairs.append((center, tokenized_text[i]))
    return pairs

pairs = generate_skipgram_pairs(words, 2)
pairs[:10]
```

```python
inputs = torch.tensor([word2idx[c] for c,_ in pairs])
targets = torch.tensor([word2idx[t] for _,t in pairs])
len(inputs)
```

```python
class SkipGramDataset(Dataset):
    def __init__(self, inputs, targets):
        self.inputs = inputs
        self.targets = targets
    def __len__(self): return len(self.inputs)
    def __getitem__(self, idx): return self.inputs[idx], self.targets[idx]

dataset = SkipGramDataset(inputs, targets)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)
```

```python
class SkipGramModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.linear = nn.Linear(embedding_dim, vocab_size)
    def forward(self, center):
        emb = self.embedding(center)
        return self.linear(emb)

model = SkipGramModel(vocab_size, 50)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

for epoch in range(5):
    total_loss=0
    for c,t in dataloader:
        optimizer.zero_grad()
        out = model(c)
        loss = criterion(out, t)
        loss.backward()
        optimizer.step()
        total_loss+=loss.item()
    print('Epoch',epoch,'Loss',total_loss)
```

```python
embeddings = model.embedding.weight.data
embeddings.shape
```

```python
import torch.nn.functional as F

def most_similar(word, topn=5):
    if word not in word2idx: return []
    vec = embeddings[word2idx[word]]
    sims = F.cosine_similarity(vec.unsqueeze(0), embeddings)
    top = sims.argsort(descending=True)[1:topn+1]
    return [idx2word[i.item()] for i in top]

most_similar('maroc')
```
